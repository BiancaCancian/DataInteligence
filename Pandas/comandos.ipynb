Editar tabela --> Pegar apenas colunas especificas

```python
produtos = vendas_df[['Produto', 'ID Loja']]
display(produtos)
```

Metodo .loc --> Pegar linha de acordo com uma condição(colunas e linhas)

```python
# Pegar 5 linhas
#display(vendas_df.loc[1:5])
```

```python
#Pegar linhas que correspondem a uma condição
vendas_norteshopping_df = vendas_df.loc[vendas_df['ID Loja'] == 'Norte Shopping']
```

```python
#pegar varias linhas e colunas usando .loc
#.loc[linhas, colunas]
vendas_norteshopping_df = vendas_df.loc[vendas_df['ID Loja'] == 'Norte Shopping', ["ID Loja", "Produto", "Quantidade"]]
display(vendas_norteshopping_df)
```

```python
#Pegar algo especifico(1 valor)
#Passar linha dele e coluna dele

print(vendas_df.loc[1, 'Produto'])
```

Adicionar 1 Coluna nova ou Linha 

```python
# a partir de uma coluna que existe
# criar coluna comissão valor_final x 0.05
vendas_df['Comissão'] = vendas_df['Valor Final'] * 0.05
display(vendas_df)
```

```python
#criar uma coluna nova com valor padrão
#Coluna de imposto preencher todos valores com 0
vendas_df.loc[:, "Imposto"] = 0
display(vendas_df)
```

```python
# Importando e lendo nova planilha
vendas_dez_df = pd.read_excel("Vendas - Dez.xlsx")
display(vendas_dez_df)
```

```python
# Adicionar vendas dez ao Vendas  
vendas_df = vendas_df.append(vendas_dez_df)
display(vendas_df)
```

 excluir linhas e colunas

```python
# 1 eixo da coluna 0 eixo da linha
vendas_df = vendas_df.drop(0, axis=0)
display(vendas_df)
```

Tratar valores vazios (Nann)

```python
#deletar linhas e colunas completamente vazias
vendas_df =  vendas_df.dropna(how='all', axis=1)
```

```python
# Se tiver pelo menos 1 valor vazio(excluir)
vandas_df =  vendas_df.dropna()
```

```python
# preencher valores vazios com a media da coluna
vendas_df['Comissão'] = vendas_df['Comissão'].fillna(vendas_df['Comissão'].mean())
display(vendas_df)
```

```python
# preencher com o ultimo valor
#ffill --> preenche com o valor que esta acima dele
vendas_df = vendas_df.ffill()
```

Calcular indicadores

```python
# quantas transações teve por loja
# value_counts --> quantos valores aparecem 
trans_lojas = vendas_df['ID Loja'].value_counts()
display(trans_lojas)
```

```python
# Faturamento de cada um dos produtos(tabela com todos os produtos, valor final somado)
# quando agrupar somar a coluna de valor final  

faturamento_produto = vendas_df.groupby('Produto').sum()
display(faturamento_produto)
```

```python
#mostrar so produto e valor final 
# Realizei primeiro o parametro que queria como retorno --> ['Produto', 'Valor Final']
faturamento_produto = vendas_df[['Produto', 'Valor Final']].groupby('Produto').sum()
display(faturamento_produto)
```

Mesclar 2 dataframes (Procurar informações de 1 dataframe no outro)

```python
gerentes_df = pd.read_excel("Gerentes.xlsx")
display(gerentes_df)
```

```python
# Fazer com que  Tabela vendas_df tenha a coluna gerentes  
vendas_df = vendas_df.merge(gerentes_df)
display(vendas_df)
```
